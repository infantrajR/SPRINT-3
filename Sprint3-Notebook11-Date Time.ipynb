# Date & Time Operations with Pandas

Date and time operations are important when working with time-based data.

Pandas provides powerful tools for working with dates, timestamps, time differences, and time zones.

In this notebook, we will learn datetime, Timestamp, date parsing, Timedelta, time zones, and extracting useful date features.

```python
import pandas as pd
from datetime import datetime
```

# 1. Datetime
Datetime represents a specific date and time.

Python provides the datetime class for working with date and time values.

Pandas also provides convenient datetime functionality for working with datasets.

```python
now = datetime.now()

print("Current date and time:")
print(now)
```

# 2. Timestamp
A Pandas Timestamp represents a specific point in time.

It is similar to Python's datetime object but provides additional functionality for data analysis.

```python
timestamp = pd.Timestamp("2026-10-08 14:30:00")

print(timestamp)
```

```python
current_timestamp = pd.Timestamp.now()

print(current_timestamp)
```

```python
print("Year:", timestamp.year)
print("Month:", timestamp.month)
print("Day:", timestamp.day)
print("Hour:", timestamp.hour)
```

# 3. Date Parsing
Date parsing means converting date strings into datetime objects.

Pandas provides pd.to_datetime() to convert strings and other date values into datetime format.

```python
dates = pd.Series([
    "2026-01-10",
    "2026-02-15",
    "2026-03-20"
])

parsed_dates = pd.to_datetime(dates)

print(parsed_dates)
```

```python
dates = pd.Series([
    "10/01/2026",
    "15/02/2026",
    "20/03/2026"
])

parsed_dates = pd.to_datetime(
    dates,
    dayfirst=True
)

print(parsed_dates)
```

```python
df = pd.DataFrame({
    "Name": ["Infy", "Prisci", "Zane"],
    "Join_Date": [
        "2026-01-10",
        "2026-02-15",
        "2026-03-20"
    ]
})

df["Join_Date"] = pd.to_datetime(df["Join_Date"])

print(df)
print(df.dtypes)
```

# 4. Timedelta
Timedelta represents a difference or duration between two dates or times.

It can be used to add or subtract days, hours, minutes, and other time units from dates.

```python
date = pd.Timestamp("2026-10-08")

new_date = date + pd.Timedelta(days=10)

print("Original date:", date)
print("After 10 days:", new_date)
```

```python
start_date = pd.Timestamp("2026-01-01")
end_date = pd.Timestamp("2026-01-15")

difference = end_date - start_date

print("Difference:", difference)
```

```python
time = pd.Timestamp("2026-10-08 10:00:00")

new_time = time + pd.Timedelta(hours=5)

print(new_time)
```

# 5. Time Zones
A time zone represents the region-specific time offset from Coordinated Universal Time (UTC).

Pandas can create timezone-aware timestamps and convert them between different time zones.

```python
utc_time = pd.Timestamp(
    "2026-10-08 10:00:00",
    tz="UTC"
)

print(utc_time)
```

```python
india_time = utc_time.tz_convert("Asia/Kolkata")

print("UTC:", utc_time)
print("India:", india_time)
```

```python
new_york_time = utc_time.tz_convert("America/New_York")

print("New York:", new_york_time)
```

# 6. Extracting Date Features
Date features are individual components extracted from a date.

Common features include year, month, day, weekday, hour, and quarter.

These features are especially useful in data analysis and machine learning.

```python
df = pd.DataFrame({
    "Date": pd.to_datetime([
        "2026-01-10",
        "2026-02-15",
        "2026-03-20",
        "2026-04-25"
    ])
})

print(df)
```

```python
df["Year"] = df["Date"].dt.year

print(df)
```

```python
df["Month"] = df["Date"].dt.month

print(df)
```

```python
df["Day"] = df["Date"].dt.day

print(df)
```

```python
df["Day_Name"] = df["Date"].dt.day_name()

print(df)
```

```python
df["Quarter"] = df["Date"].dt.quarter

print(df)
```

```python
df = pd.DataFrame({
    "Date": pd.to_datetime([
        "2026-01-10",
        "2026-02-15",
        "2026-03-20",
        "2026-04-25"
    ])
})

df["Year"] = df["Date"].dt.year
df["Month"] = df["Date"].dt.month
df["Day"] = df["Date"].dt.day
df["Day_Name"] = df["Date"].dt.day_name()
df["Quarter"] = df["Date"].dt.quarter

print(df)
```

# Practical Example
Let's create a sales dataset with dates and extract useful date features for analysis.

```python
sales = pd.DataFrame({
    "Date": pd.to_datetime([
        "2026-01-05",
        "2026-01-15",
        "2026-02-10",
        "2026-02-25",
        "2026-03-12"
    ]),
    "Sales": [10000, 15000, 12000, 18000, 20000]
})

sales["Year"] = sales["Date"].dt.year
sales["Month"] = sales["Date"].dt.month
sales["Month_Name"] = sales["Date"].dt.month_name()
sales["Day_Name"] = sales["Date"].dt.day_name()

print(sales)
```

```python
monthly_sales = sales.groupby(
    "Month"
)["Sales"].sum()

print(monthly_sales)
```
