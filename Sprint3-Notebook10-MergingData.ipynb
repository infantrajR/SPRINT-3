# Combining Datasets with Pandas

Combining datasets is an important part of data analysis.

Pandas provides several methods for combining DataFrames, including concat(), merge(), join(), and combine_first().

In this notebook, we will learn how these methods work and when to use each one.

```python
import pandas as pd
```

```python
students1 = pd.DataFrame({
    "Student_ID": [1, 2, 3],
    "Name": ["Infy", "Prisci", "Zane"],
    "Marks": [95, 90, 85]
})

students2 = pd.DataFrame({
    "Student_ID": [4, 5, 6],
    "Name": ["Deepan", "Arun", "Akira"],
    "Marks": [93, 78, 75]
})

print("Students 1:")
print(students1)

print("\nStudents 2:")
print(students2)
```

# 1. concat()
The concat() function combines DataFrames along rows or columns.

It is useful when datasets have the same or similar structure.

By default, concatenation happens along rows.

```python
combined = pd.concat(
    [students1, students2],
    ignore_index=True
)

print(combined)
```

```python
marks = pd.DataFrame({
    "Student_ID": [1, 2, 3],
    "Marks": [85, 90, 78]
})

grades = pd.DataFrame({
    "Grade": ["B", "A", "C"]
})

result = pd.concat(
    [marks, grades],
    axis=1
)

print(result)
```

# 2. merge()
The merge() function combines DataFrames based on one or more common columns.

It is similar to SQL JOIN operations.

It is useful when two datasets contain related information connected by a common key.

```python
students = pd.DataFrame({
    "Student_ID": [1, 2, 3, 4],
    "Name": ["Infy", "Prisci", "Zane", "Deepan"]
})

marks = pd.DataFrame({
    "Student_ID": [1, 2, 3, 4],
    "Marks": [95, 90, 85, 93]
})
merged = pd.merge(
    students,
    marks,
    on="Student_ID"
)

print(merged)
```

```python
inner_merge = pd.merge(
    students,
    marks,
    on="Student_ID",
    how="inner"
)

print(inner_merge)
```

```python
left_merge = pd.merge(
    students,
    marks,
    on="Student_ID",
    how="left"
)

print(left_merge)
```

```python
outer_merge = pd.merge(
    students,
    marks,
    on="Student_ID",
    how="outer"
)

print(outer_merge)
```

# 3. join()
The join() method combines DataFrames based primarily on their indexes.

It is useful when related data is already aligned using the same index.

```python
student_info = pd.DataFrame({
    "Name": ["Infy", "prisci", "Zane"]
}, index=[1, 2, 3])

student_marks = pd.DataFrame({
    "Marks": [95, 90, 85]
}, index=[1, 2, 3])

joined = student_info.join(student_marks)

print(joined)          
```

```python
student_marks2 = pd.DataFrame({
    "Marks": [95, 90, 93]
}, index=[1, 2, 4])

joined = student_info.join(
    student_marks2,
    how="left"
)

print(joined)
```

# 4. append()
The DataFrame.append() method was previously used to add rows to a DataFrame.

However, DataFrame.append() was deprecated and removed from modern versions of Pandas.

The recommended alternative is pd.concat().

```python
combined = pd.concat(
    [students1, students2],
    ignore_index=True
)

print(combined)
```

# 5. combine_first()
The combine_first() method fills missing values in one DataFrame using values from another DataFrame.

It is useful when we have two datasets containing complementary information.

```python
df1 = pd.DataFrame({
    "Name": ["Infy", "prisci", "Zane"],
    "Marks": [95, None, 85]
})

print(df1)
```

```python
df2 = pd.DataFrame({
    "Name": ["Infy", "prisci", "Zane"],
    "Marks": [None, 90, 85]
})

print(df2)
```

```python
result = df1.combine_first(df2)

print(result)
```

# Practical Example
Let's combine student information and marks from two different datasets.

We will use merge() because both datasets contain a common Student_ID.

```python
student_details = pd.DataFrame({
    "Student_ID": [101, 102, 103, 104],
    "Name": ["Infy", "Prisci", "Zane", "Deepan"],
    "City": ["Chennai", "Chennai", "Karur", "Salem"]
})

student_marks = pd.DataFrame({
    "Student_ID": [101, 102, 103, 104],
    "Marks": [95, 90, 85, 92]
})

final_data = pd.merge(
    student_details,
    student_marks,
    on="Student_ID",
    how="inner"
)

print(final_data)
```

# Comparison of Combining Methods

Method	Main Purpose

concat()	Combines DataFrames along rows or columns

merge()	Combines DataFrames using common columns/keys

join()	Combines DataFrames mainly using indexes

append()	Old row-appending method; deprecated/removed

combine_first()	Fills missing values using another DataFrame
