# Missing Values

```python
import pandas as pd
import numpy as np
```

Create a Sample Dataset

```python
students = pd.DataFrame({
    "Student_ID": [101, 102, 103, 104, 105, 106],
    "Name": ["Infy", "Prisci", "Zane", "Arun", "Deepan", "Akira"],
    "Age": [33, 29, np.nan, 21, 24, np.nan],
    "Department": ["AI", "ML", "DS", "AI", np.nan, "ML"],
    "Marks": [95, np.nan, 88, 76, np.nan, 91],
    "Attendance": [92.5, 88.0, np.nan, 75.5, 90.0, np.nan]
})

print(students)
```

# 1.isnull()
isnull() identifies missing values.

It returns:

True → Missing

False → Not missing

```python
print(students.isnull())
```

```python
# Count Missing Values

print(students.isnull().sum())
```

```python
# Total Missing Values

total_missing = students.isnull().sum().sum()

print(total_missing)
```

# Real-World Example
Before performing EDA or building a machine learning model, we should always check for null values, which gives us an initial understanding of data quality.

# 2.notnull()
notnull() identifies values that are not missing.

It returns:

True → Value exists

False → Value is missing

```python
print(students.notnull())
```

```python
# Count Available Values

print(students.notnull().sum())
```

Real-World Example

Suppose a company wants to analyze only customers who have provided their phone number

# 3.dropna()
dropna() removes rows or columns containing missing values.

This is useful when missing data cannot be reliably recovered.

```python
# Drop Rows Containing Any Missing Value

clean_students = students.dropna()

print(clean_students)
```

dropna(axis=1)
We can remove columns containing missing values.

```python
result = students.dropna(axis=1)

print(result)
```

# Real-World Example
In a loan approval dataset, suppose these fields are mandatory:

Income

Age

Credit_Score

# 4.fillna()
fillna() replaces missing values with a specified value.

This is one of the most commonly used missing-value techniques.

```python
# Fill with a Constant

students["Marks"].fillna(0)
```

```python
# Fill with Mean

mean_marks = students["Marks"].mean()

students["Marks"] = students["Marks"].fillna(mean_marks)
print(students)
```

```python
# Fill with Median

median_age = students["Age"].median()

students["Age"] = students["Age"].fillna(median_age)
print (students)
```

```python
# Fill Categorical Values with Mode

mode_department = students["Department"].mode()[0]

students["Department"] = students["Department"].fillna(
    mode_department
)

print(students)
```

# 5.interpolate()
interpolate() estimates missing values based on surrounding values.

It is especially useful for ordered or time-series data.

```python
temperature = pd.DataFrame({
    "Time": [
        "10:00",
        "10:01",
        "10:02",
        "10:03",
        "10:04"
    ],
    "Temperature": [
        28.0,
        28.5,
        np.nan,
        29.5,
        30.0
    ]
})

print(temperature)
```

The missing value can be estimated between 28.5 and 29.5.

# Linear Interpolation

```python
temperature["Temperature"] = (
    temperature["Temperature"]
    .interpolate()
)

print(temperature)
```

The missing value was estimated as:

(28.5 + 29.5) / 2 = 29.0

# Real World Example:
Interpolation can help preserve the continuity of time-series data before feeding it into a model.# 

# 6.Missing Value Handling Strategies
There is no single best method for handling missing values.

The correct approach depends on:

Why the data is missing

How much data is missing

Data type

Distribution

Business meaning

ML 

# Strategy 1 – Drop the Rows
Use:

df.dropna()

# When to use?
When:

Only a small number of rows are missing

The missing information is critical

Removing the rows will not significantly reduce the dataset

# Example:
100,000 rows --> 50 missing rows

Dropping 50 rows may be acceptable.

# Strategy 2 – Drop the Column
# Use:

df.dropna(axis=1)

But this is usually appropriate only when a column has very high missingness and provides little useful information.

# Example:

Customer_ID --> 0% missing

Age --> 5% missing

Income --> 8% missing

Middle_Name --> 95% missing

It may make sense to remove Middle_Name.

# Strategy 3 – Mean Imputation
# Use:

df["Age"] = df["Age"].fillna(df["Age"].mean())

Suitable for:

Numerical data

Relatively symmetric distributions

Low/moderate missingness

# Limitation:

Mean can be strongly affected by outliers.

# Strategy 4 – Median Imputation
Use:

df["Income"] = df["Income"].fillna(df["Income"].median())

# Suitable for:

Numerical data

Skewed distributions

Data containing outliers

Median is generally more robust than mean to extreme values.

# Strategy 5 – Mode Imputation
# Use:

df["Department"] = df["Department"].fillna(df["Department"].mode()[0])

# Suitable for:

Categorical variables

Small amounts of missing data

Strategy 6 – Forward Fill

Forward fill uses the previous available value.

df["Temperature"] = df["Temperature"].ffill()

# Example:

# Strategy 7 – Backward Fill
Backward fill uses the next available value.

df["Temperature"] = df["Temperature"].bfill()
