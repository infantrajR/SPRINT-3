# Notebook 5 – Reading & Writing Data

# 1. CSV Files
Definition
A CSV (Comma-Separated Values) file stores tabular data in rows and columns separated by commas. It is one of the most commonly used formats for datasets. CSV files are simple, lightweight, and supported by many applications.

# Real-Life Example
A company can store customer names, product details, and sales records in CSV files. Employees can open the same file in Excel or process it using Python. This makes CSV useful for simple data sharing and reporting.

# AI/ML Example
In AI/ML, many datasets are provided in CSV format. Pandas can read the dataset into a DataFrame for data cleaning, preprocessing, feature selection, and model training. Processed data can also be written back to another CSV file.

# Explanation
pd.read_csv() reads an existing CSV file and loads it into a DataFrame. df.to_csv() writes the DataFrame into a new CSV file. index=False prevents the DataFrame index from being saved as an additional column.

# Python Implementation

```python
import pandas as pd

# Reading
df = pd.read_csv("students.csv")
print(df)

# Writing
df.to_csv("students_output.csv", index=False)
```

# 2. Excel Files
Definition
An Excel file stores tabular data in one or more worksheets and commonly uses .xlsx or .xls extensions. Pandas can read Excel worksheets into DataFrames and write DataFrames back to Excel files. Excel is widely used for business reports and structured records.

# Real-Life Example
A company may maintain employee names, departments, salaries, and attendance in an Excel file. Pandas can read the employee sheet, process the records, and save the updated information into another Excel file.

# AI/ML Example
In AI/ML, client or business datasets are often received in Excel format. Pandas loads the required worksheet into a DataFrame so the data can be cleaned and prepared for model training. Processed results can also be exported to Excel.

# Explanation
pd.read_excel() reads an existing Excel file into a DataFrame. df.to_excel() writes the DataFrame into a new Excel file. Working with .xlsx files normally requires the openpyxl library.

# Python Implementation

```python
import pandas as pd

# Reading
df = pd.read_excel("students.xlsx")
print(df)

# Writing
df.to_excel("students_output.xlsx", index=False)
```

# 3. JSON Files
Definition
JSON (JavaScript Object Notation) stores structured data using key-value pairs. It is commonly used by APIs, websites, and applications for exchanging data. Pandas can convert suitable JSON records into a DataFrame.

# Real-Life Example
A web application may store user names, ages, and cities in JSON format. Python can read these records, process the information, and save the updated data into another JSON file.

# AI/ML Example
In AI/ML, data collected from REST APIs is often received in JSON format. Pandas can read the JSON data into a DataFrame for preprocessing and analysis. Cleaned or transformed records can then be written back to JSON.

# Explanation
pd.read_json() reads JSON data and converts it into a DataFrame. df.to_json() writes the DataFrame into JSON format. orient="records" stores each DataFrame row as a separate JSON record.

# Python Implementation

```python
import pandas as pd

# Reading
df = pd.read_json("students.json")
print(df)

# Writing
df.to_json("students_output.json", orient="records", indent=4)
```

# 4. Parquet Files
Definition
Parquet is a column-based binary storage format designed for efficient data storage and processing. It usually uses less storage than CSV and is useful for large datasets. Pandas supports both reading and writing Parquet files.

# Real-Life Example
A large company may store millions of transaction records in Parquet format. The format can reduce storage requirements and allow analytical systems to work efficiently with selected columns.

# AI/ML Example
In AI/ML, large processed datasets and feature tables are often stored as Parquet files. Pandas can load them efficiently for preprocessing and model training. Processed datasets can also be written back to Parquet for later use.

# Explanation
pd.read_parquet() reads an existing Parquet file into a DataFrame. df.to_parquet() writes the DataFrame into a new Parquet file. Pandas normally requires pyarrow or fastparquet for Parquet support.

# Python Implementation

```python
import pandas as pd

# Reading
df = pd.read_parquet("students.parquet")
print(df)

# Writing
df.to_parquet("students_output.parquet", index=False)
```

# 5. Pickle Files
Definition
A Pickle file stores Python objects in binary format and commonly uses the .pkl extension. It can preserve objects such as Pandas DataFrames so they can be loaded again later. Pickle is mainly intended for Python-based workflows.

# Real-Life Example
A company can save a processed DataFrame as a Pickle file after completing time-consuming data preparation. Employees can later reload the saved object instead of repeating all the processing steps.

# AI/ML Example
In AI/ML, Pickle files can store preprocessed DataFrames and other Python objects for reuse. Python's pickle or joblib tools are also commonly used to persist trained models. This can make prediction and deployment workflows easier.

# Explanation
pd.read_pickle() loads a Pickle object into Python, while df.to_pickle() saves a DataFrame as a Pickle file. Pickle files should only be loaded from trusted sources because unsafe Pickle files can execute malicious code.

# Python Implementation

```python
import pandas as pd

# Reading
df = pd.read_pickle("students.pkl")
print(df)

# Writing
df.to_pickle("students_output.pkl")
```

# 6. SQL Tables
Definition
A SQL table stores structured data inside a relational database using rows and columns. Pandas can retrieve records from SQL tables and convert them into DataFrames. It can also write DataFrame records back into database tables.

# Real-Life Example
A bank may store customer and transaction information in SQL database tables. Pandas can retrieve selected records for analysis and write processed results into another table when required.

# AI/ML Example
In AI/ML, training data is often stored in relational databases. Pandas can execute a SQL query and load the required records into a DataFrame for preprocessing and feature engineering. Prepared data can also be written to a SQL table.

# Explanation
sqlite3.connect() opens a connection to the SQLite database. pd.read_sql() reads query results into a DataFrame, while df.to_sql() writes the DataFrame into a database table. The connection should be closed after the operations are completed.

# Python Implementation

```python
import pandas as pd
import sqlite3

# Connect to database
conn = sqlite3.connect("students.db")

# Reading
df = pd.read_sql("SELECT * FROM students", conn)
print(df)

# Writing
df.to_sql("students_output", conn, if_exists="replace", index=False)

# Close connection
conn.close()
```

# 7. Reading Multiple Files
Definition
Reading multiple files means loading data from several files and combining or processing them together. This is useful when information is divided into daily, monthly, departmental, or branch-wise files. Pandas can automate the complete process.

# Real-Life Example
A company may store monthly sales in separate CSV files inside one folder. Instead of opening each report manually, Python can find all the files and combine them into one yearly sales dataset.

# AI/ML Example
In AI/ML, a large dataset may be divided across multiple files or sources. Pandas can read each file and combine the records into one DataFrame before preprocessing and model training. The combined dataset can then be written to a new file.

# Explanation
glob.glob() finds files that match a pattern such as data/*.csv. pd.read_csv() reads each file, while pd.concat() combines the DataFrames into one. ignore_index=True creates a new continuous index after combining the records.

# Python Implementation

```python
import pandas as pd
import glob

# Reading multiple CSV files
files = glob.glob("data/*.csv")

df = pd.concat(
    [pd.read_csv(file) for file in files],
    ignore_index=True
)

print(df)
```

# 8. File Path Handling
Definition
File path handling means specifying and managing the correct location of files and folders in a program. A path tells Python where a file should be read from or written to. Correct path handling helps avoid FileNotFoundError and keeps projects organized.

# Real-Life Example
A company may store reports inside folders such as Sales/2026/October/. The correct file path allows the program to locate the required report automatically and save output files in the correct folder.

# AI/ML Example
In AI/ML, datasets, trained models, notebooks, and prediction results are often stored in separate folders. Proper file paths ensure that the correct resources are loaded during training and testing. They also help save processed outputs in organized locations.

# Explanation
Path from pathlib provides a clean way to create file and folder paths. The / operator can join folders and filenames without manually handling slash styles. The same path can then be passed directly to Pandas reading and writing functions.

# Python Implementation

```python
import pandas as pd
from pathlib import Path

# File paths
input_path = Path("data") / "students.csv"
output_path = Path("data") / "students_output.csv"

# Reading
df = pd.read_csv(input_path)
print(df)

# Writing
df.to_csv(output_path, index=False)
```
