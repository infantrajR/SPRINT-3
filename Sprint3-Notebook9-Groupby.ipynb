# Grouping and Aggregation with Pandas

Grouping and aggregation are important techniques used to summarize and analyze data.

Pandas provides functions such as groupby(), agg(), transform(), and filter() for grouping and analyzing data.

We will also learn how to create Pivot Tables and Crosstab tables.

```python
import pandas as pd
import numpy as np
```

```python
data = {
    "Name": [
        "Infy", "Prisci", "Zane", "Arun",
        "Deepan", "Zane", "Akira", "Raja"
    ],
    "Department": [
        "IT", "HR", "IT", "Finance",
        "HR", "IT", "Finance", "HR"
    ],
    "City": [
        "Trichy", "Chennai", "Salem", "Bangalore",
        "Chennai", "Trichy", "Bangalore", "Chennai"
    ],
    "Marks": [85, 90, 78, 92, 88, 76, 95, 89],
    "Salary": [
        50000, 45000, 55000, 60000,
        48000, 52000, 65000, 47000
    ]
}

df = pd.DataFrame(data)

print(df)
```

# 1. groupby()
The groupby() method groups rows based on one or more columns.

It is useful for calculating statistics separately for each group.

For example, we can group employees by department and calculate their average salary.

```python
department_salary = df.groupby("Department")["Salary"].mean()

print(department_salary)
```

```python
city_marks = df.groupby("City")["Marks"].mean()

print(city_marks)
```

```python
result = df.groupby(
    ["Department", "City"]
)["Salary"].mean()

print(result)
```

# 2. agg()
The agg() method allows us to apply multiple aggregation functions at the same time.

Common aggregation functions include mean, sum, min, max, and count.

```python
result = df.groupby("Department")["Salary"].agg(
    ["mean", "sum", "min", "max", "count"]
)

print(result)
```

```python
result = df.groupby("Department").agg({
    "Salary": ["mean", "max"],
    "Marks": ["mean", "min", "max"]
})

print(result)
```

# 3. transform()
The transform() method performs a calculation for each group but returns a result with the same number of rows as the original DataFrame.

It is useful when we want to compare individual rows with their group-level statistics.

```python
df["Department_Avg_Salary"] = df.groupby(
    "Department"
)["Salary"].transform("mean")

print(df)
```

```python
df["Salary_Difference"] = (
    df["Salary"] - df["Department_Avg_Salary"]
)

print(df)
```

# 4. filter()
The filter() method can be used with groupby() to keep or remove entire groups based on a condition.

For example, we can keep only departments whose average salary is greater than a specified value.

```python
filtered = df.groupby("Department").filter(
    lambda group: group["Salary"].mean() > 50000
)

print(filtered)
```

```python
filtered = df.groupby("Department").filter(
    lambda group: len(group) >= 2
)

print(filtered)
```

# 5. Pivot Tables
A Pivot Table summarizes data by grouping it across rows and columns.

Pandas provides the pivot_table() function to create Pivot Tables.

Pivot Tables are useful for quickly comparing values across different categories.

```python
pivot = pd.pivot_table(
    df,
    values="Salary",
    index="Department",
    columns="City",
    aggfunc="mean"
)

print(pivot)
```

```python
pivot_marks = pd.pivot_table(
    df,
    values="Marks",
    index="Department",
    aggfunc="mean"
)

print(pivot_marks)
```

# 6. Crosstab
A Crosstab, or cross-tabulation, summarizes the relationship between two or more categorical variables.

Pandas provides the pd.crosstab() function.

It is commonly used to calculate frequency counts.

```python
cross_table = pd.crosstab(
    df["Department"],
    df["City"]
)

print(cross_table)
```

```python
cross_table = pd.crosstab(
    df["Department"],
    df["City"],
    values=df["Marks"],
    aggfunc="mean"
)

print(cross_table)
```

# Practical Example
Let's analyze employee salary data by department.

We will calculate the total salary, average salary, and number of employees in each department.

```python
summary = df.groupby("Department").agg(
    Average_Salary=("Salary", "mean"),
    Total_Salary=("Salary", "sum"),
    Employee_Count=("Name", "count")
)

print(summary)
```

# Grouping Methods Comparison

Method	Purpose

groupby()	Groups data based on one or more columns

agg()	Applies multiple aggregation functions

transform()	Returns group calculations with the same number of rows

filter()	Keeps or removes complete groups

pivot_table()	Creates summarized tables using rows and columns

crosstab()	Shows relationships and frequency counts between categories
