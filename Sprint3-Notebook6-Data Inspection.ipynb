# Data Inspection with Pandas

Data inspection is an important step in data analysis.

It helps us understand the structure, size, data types, statistical properties, and unique values in a dataset.

In this notebook, we will learn different Pandas functions and attributes used to inspect and understand data.

```python
import pandas as pd
import numpy as np
```

```python
data = {
    "Name": ["Infant", "Prisci", "Deepan", "Arun", "Zane", "Jerin", "Akira", "Avril"],
    "Age": [33, 29, 34, 33, 21, 27, 22, 25],
    "City": [
        "Hyderabad",
        "Chennai",
        "Hyderabad",
        "Bangalore",
        "Chennai",
        "Hyderabad",
        "Bangalore",
        "Hyderabad"
    ],
    "Marks": [85, 90, 78, 92, 88, 76, 95, 89],
    "Passed": [True, True, True, True, True, True, True, True]
}

df = pd.DataFrame(data)

print(df)
```

# 1. head()
The head() method displays the first few rows of a DataFrame.

By default, it displays the first 5 rows.

It is useful for quickly checking the beginning of a dataset.

```python
print(df.head())
```

```python
print(df.head(3))
```

# 2. tail()
The tail() method displays the last few rows of a DataFrame.

By default, it displays the last 5 rows.

It is useful for checking the end of a dataset.

```python
print(df.tail())
```

# 3. sample()
The sample() method randomly selects rows from a DataFrame.

It is useful when we want to look at random records instead of only the beginning or end of the dataset.

```python
print(df.sample(3))
```

```python
print(df.sample(3, random_state=42))
```

# 4. info()
The info() method provides a concise summary of a DataFrame.

It shows the number of rows, columns, column names, non-null values, and data types.

```python
  df.info()   
```

# 5. describe()
The describe() method generates descriptive statistics for numerical columns.

It provides values such as count, mean, standard deviation, minimum, maximum, and quartiles.

```python
print(df.describe())
```

```python
print(df["Marks"].describe())
```

# 6. shape
The shape attribute returns the number of rows and columns in a DataFrame.

The result is returned as a tuple:

(rows, columns)

```python
print("Shape:", df.shape)
```
