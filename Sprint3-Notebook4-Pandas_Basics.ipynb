# 1. What is Pandas?
Definition
Pandas is a Python library used for working with structured data such as tables, rows, and columns. It provides simple tools for organizing, analyzing, cleaning, and modifying data. It is widely used in data analysis and AI/ML projects.

# Real-Life Example
A shop can store product names, prices, quantities, and sales in a table. Pandas makes it easy to view the table, calculate values, and update the records.

# AI/ML Example
In AI/ML, Pandas is commonly used to load, explore, clean, and prepare datasets before model training. It makes tabular data easy to inspect and modify. This makes it easier to convert raw tabular information into clean features and target data for machine-learning workflows.

# Explanation
Pandas mainly provides two data structures: Series for one-dimensional data and DataFrame for two-dimensional tabular data. These Pandas operations make structured datasets easier to inspect, organize, and prepare for further analysis.

# Python Implementation

```python
import pandas as pd

data = {
    "Name": ["Asha", "Ravi", "Kiran"],
    "Marks": [78, 85, 92]
}

df = pd.DataFrame(data)
print(df)
```

# 2. Why Pandas?
Definition
Pandas makes it easy to organize, clean, transform, analyze, read, and write structured data. It provides many built-in functions for handling tabular datasets efficiently. This reduces the amount of manual code needed for data processing.

# Real-Life Example
If a company has thousands of customer records, Pandas can quickly filter customers, calculate averages, and handle missing information. This is much easier than checking every record manually.

# AI/ML Example
In AI/ML, raw datasets usually need preprocessing before they can be given to a model. Pandas helps handle missing values, select features, filter records, and prepare training data. This makes it easier to convert raw tabular information into clean features and target data for machine-learning workflows.

# Explanation
Pandas provides simple functions for many common data operations. It reduces the amount of manual Python code required when working with tables. These Pandas operations make structured datasets easier to inspect, organize, and prepare for further analysis.

# Python Implementation

```python
import pandas as pd

data = {
    "Name": ["Asha", "Ravi", "Kiran"],
    "Marks": [78, 85, 92]
}

df = pd.DataFrame(data)

print("Average Marks:", df["Marks"].mean())
print(df[df["Marks"] > 80])
```

# 3. Series
Definition
A Series is a one-dimensional Pandas data structure that stores a sequence of values with labels called indexes. It is similar to a single column in a table. A Series can store numbers, text, or other types of data.

# Real-Life Example
A student's marks from different tests can be stored as a Pandas Series. Each test name can be used as an index, making the corresponding mark easy to access.

# AI/ML Example
In AI/ML, a Series can represent a single feature or target column from a dataset. For example, the Age column or model target labels can be accessed as a Series. This makes it easier to convert raw tabular information into clean features and target data for machine-learning workflows.

# Explanation
A Series contains values and an associated index. It can be created using pd.Series() from lists, arrays, or dictionaries. These Pandas operations make structured datasets easier to inspect, organize, and prepare for further analysis.

# Python Implementation

```python
import pandas as pd

marks = pd.Series([75, 82, 90, 68],
                  index=["Test1", "Test2", "Test3", "Test4"])

print(marks)
```

# 4. DataFrame
Definition
A DataFrame is a two-dimensional Pandas data structure that stores data in rows and columns. It is similar to an Excel sheet or database table. Different columns can contain different types of data.

# Real-Life Example
A student table can contain Name, Age, Course, and Marks as separate columns. Each row represents the complete information of one student.

# AI/ML Example
In AI/ML, datasets are commonly stored as DataFrames where rows represent samples and columns represent features. This makes preprocessing and feature selection easier. This makes it easier to convert raw tabular information into clean features and target data for machine-learning workflows.

# Explanation
A DataFrame is similar to a spreadsheet or database table. Each column can contain a different data type. These Pandas operations make structured datasets easier to inspect, organize, and prepare for further analysis.

# Python Implementation

```python
import pandas as pd

data = {
    "Name": ["Arun", "Meena", "Rahul"],
    "Age": [21, 22, 20],
    "Marks": [80, 88, 76]
}

df = pd.DataFrame(data)
print(df)
```

# 5. Creating DataFrames
Definition
Creating a DataFrame means organizing data into a two-dimensional table with rows and columns. Pandas can create DataFrames from dictionaries, lists, NumPy arrays, and other sources. Column names help describe the stored information.

# Real-Life Example
Employee names, departments, and salaries can be combined into one DataFrame. Each employee becomes a row while Name, Department, and Salary become columns.

```python
import pandas as pd

data = {
    "Product": ["Laptop", "Phone", "Tablet"],
    "Price": [55000, 30000, 22000],
    "Stock": [10, 25, 15]
}

df = pd.DataFrame(data)
print(df)
```

# 6. Data Types
Definition
Data types describe the kind of values stored in each DataFrame column. Common Pandas data types include integer, float, object, and boolean. Checking data types helps us understand how each column can be processed.

# Real-Life Example
In an employee table, Age may be stored as an integer, Salary as a float, Name as text, and Active Status as a boolean. Each column therefore has a suitable data type.

# AI/ML Example
In AI/ML, checking data types is important before preprocessing because models mainly work with numerical values. Categorical or incorrectly stored columns may need conversion. This makes it easier to convert raw tabular information into clean features and target data for machine-learning workflows.

# Explanation
The dtypes attribute displays the data type of every column. Individual columns can also be converted using astype(). These Pandas operations make structured datasets easier to inspect, organize, and prepare for further analysis.

# Python Implementation

```python
import pandas as pd

data = {
    "Name": ["Asha", "Ravi", "Kiran"],
    "Age": [21, 23, 22],
    "Score": [85.5, 90.0, 78.5]
}

df = pd.DataFrame(data)

print(df)
print("\nData Types:")
print(df.dtypes)
```

# 7. Index
Definition
An index is a label used to identify each row in a Pandas Series or DataFrame. Pandas creates numerical indexes automatically if we do not provide our own. Custom indexes can also be used to identify records more clearly.

# Real-Life Example
A student ID such as S101, S102, and S103 can be used as the index. This allows a particular student's row to be identified easily using the student ID.

# AI/ML Example
In AI/ML, indexes help identify and select specific observations in a dataset. They can also help keep records aligned while manipulating data. This makes it easier to convert raw tabular information into clean features and target data for machine-learning workflows.

# Explanation
By default, Pandas creates indexes starting from 0. We can also assign our own meaningful index values. These Pandas operations make structured datasets easier to inspect, organize, and prepare for further analysis.

# Python Implementation

```python
import pandas as pd

data = {
    "Name": ["Asha", "Ravi", "Kiran"],
    "Marks": [80, 75, 92]
}

df = pd.DataFrame(data, index=["S101", "S102", "S103"])

print(df)
print("\nIndex:", df.index)
```

# 8. Columns
Definition
Columns are the named vertical sections of a DataFrame that represent different variables or attributes. Each column contains values related to one type of information. Column names make the dataset easier to understand and access.

# Real-Life Example
In a customer table, Name, Age, City, and Purchase Amount can be separate columns. Each column stores one specific type of customer information.

# AI/ML Example
In AI/ML, DataFrame columns usually represent input features and target variables. We can select required columns and remove unnecessary ones before model training. This makes it easier to convert raw tabular information into clean features and target data for machine-learning workflows.

# Explanation
The columns attribute returns all column names. A specific column can be selected using its name. These Pandas operations make structured datasets easier to inspect, organize, and prepare for further analysis.

# Python Implementation

```python
import pandas as pd

data = {
    "Name": ["Asha", "Ravi", "Kiran"],
    "Age": [21, 23, 22],
    "Marks": [80, 75, 92]
}

df = pd.DataFrame(data)

print("Columns:", df.columns)
print("\nMarks Column:")
print(df["Marks"])
```

# 9. Reading Data
Definition
Reading data means loading information from an external file or source into Pandas. Pandas can read formats such as CSV, Excel, and JSON. The loaded data is usually stored in a DataFrame for further processing.

# Real-Life Example
A company may have customer details saved in a CSV file. Pandas can read that file and display the customer records as rows and columns for analysis.

# AI/ML Example
In AI/ML, datasets are often stored in CSV, Excel, JSON, or other files. Pandas reads these files into DataFrames so they can be cleaned and prepared for model training. This makes it easier to convert raw tabular information into clean features and target data for machine-learning workflows.

# Explanation
Common functions include pd.read_csv(), pd.read_excel(), and pd.read_json(). The correct function depends on the file format. These Pandas operations make structured datasets easier to inspect, organize, and prepare for further analysis.

# Python Implementation

```python
import pandas as pd
from io import StringIO

csv_data = """Name,Age,Marks
Asha,21,80
Ravi,23,75
Kiran,22,92
"""

df = pd.read_csv(StringIO(csv_data))

print(df)
```

# 10. Writing Data
Definition
Writing data means saving a Pandas DataFrame into an external file. Pandas can save data in formats such as CSV, Excel, and JSON. This allows processed or updated data to be stored and reused later.

# Real-Life Example
After updating employee records, a company can save the cleaned DataFrame into a new CSV or Excel file. The saved file can then be shared or used again later.

# AI/ML Example
In AI/ML, cleaned or transformed datasets can be saved after preprocessing so they can be reused later. Predictions and analysis results can also be exported using Pandas. This makes it easier to convert raw tabular information into clean features and target data for machine-learning workflows.

# Explanation
Common methods include to_csv(), to_excel(), and to_json(). index=False is often used when the DataFrame index does not need to be saved as a separate column. These Pandas operations make structured datasets easier to inspect, organize, and prepare for further analysis.

# Python Implementation

```python
import pandas as pd

data = {
    "Name": ["Asha", "Ravi", "Kiran"],
    "Marks": [80, 75, 92]
}

df = pd.DataFrame(data)

df.to_csv("student_marks.csv", index=False)

print("Data saved to student_marks.csv")
```
