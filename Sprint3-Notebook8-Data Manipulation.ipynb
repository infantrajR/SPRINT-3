# Notebook 8 – Data Manipulation

Sample DataFrame

```python
import pandas as pd

# Sample dataset used throughout this notebook
data = {
    "Name": ["Infy", "Prisci", "Zane", "Arun", "Deepan", "Akira", "Jeni", "Raja"],
    "Age": [31, 29, 20, 33, 31, 24, 22, 20],
    "Marks": [92, 86, 71, 88, 79, 65, 79, 86],
    "City": ["Chennai", "Erode", "Salem", "Bengaluru",
             "Salem", "Mysuru", "karur", "Mysuru"]
}

df = pd.DataFrame(data)
df
```

# 1. Rename Columns
Definition
Renaming columns means changing one or more column names in a DataFrame. Pandas provides the rename() method to give columns clearer or more meaningful names. Good column names make datasets easier to understand and work with.

# Real-Life Example
A company may receive a sales dataset with short names such as Cust_Name or Amt. An analyst can rename them to Customer_Name and Amount so reports are easier to understand. Clear names also reduce confusion when several employees work with the same data.

# AI/ML Example
In AI/ML, datasets may contain unclear or inconsistent feature names. Renaming columns makes preprocessing code easier to read and helps maintain a clean feature pipeline. It is especially useful before feature engineering and model training.

# Explanation
df.rename() can rename selected columns by passing a dictionary to the columns parameter. The dictionary contains the old column name as the key and the new name as the value. The changed DataFrame can be stored in a new variable or updated using inplace=True.

# Python Implementation

```python
renamed_df = df.rename(columns={
    "Marks": "Score",
    "City": "Location"
})

print(renamed_df)
```

# 2. Drop Rows
Definition
Dropping rows means removing unwanted records from a DataFrame. Pandas provides the drop() method to remove rows using their index labels. This is useful when certain observations are incorrect, unnecessary, or should not be included in analysis.

# Real-Life Example
A company may discover that one transaction record was entered incorrectly. The analyst can remove that specific row using its index instead of rebuilding the complete dataset. This keeps the remaining valid records unchanged.

# AI/ML Example
In AI/ML, rows may be dropped when they contain invalid observations, unwanted samples, or data that cannot be used for training. Removing unsuitable records can improve data quality. However, rows should be removed carefully to avoid losing useful information.

# Explanation
df.drop(index=...) removes rows using their index labels. We can provide one index or a list of indexes when several rows need to be removed. By default, drop() returns a new DataFrame unless inplace=True is used.

# Python Implementation

```python
# Drop row with index 2
dropped_rows = df.drop(index=2)

print(dropped_rows)
```

# 3. Drop Columns
Definition
Dropping columns means removing unnecessary features or fields from a DataFrame. Pandas can remove columns using drop() with the columns parameter. This helps keep only the information required for analysis or further processing.

# Real-Life Example
A company dataset may contain an internal reference column that is not required in a report. The analyst can remove that column before sharing or analyzing the data. This makes the dataset smaller and more focused.

# AI/ML Example
In AI/ML, some columns may be irrelevant, redundant, or unsuitable for model training. Dropping unnecessary features can simplify preprocessing and reduce noise. For example, identifiers such as customer IDs may sometimes be excluded from model inputs.

# Explanation
df.drop(columns=...) removes one or more columns by name. A single column name or a list of column names can be supplied. The original DataFrame remains unchanged unless the result is assigned back or inplace=True is used.

# Python Implementation

```python
# Drop the City column
dropped_columns = df.drop(columns=["City"])

print(dropped_columns)
```

# 4. Insert Columns
Definition
Inserting columns means adding a new column at a specific position in a DataFrame. Pandas provides the insert() method where we specify the column position, name, and values. This gives more control than simply adding a column at the end.

# Real-Life Example
A company may want to add a Grade column next to the Marks column in a student report. insert() allows the new column to be placed exactly where required. This makes the final report easier to read and organize.

# AI/ML Example
In AI/ML, new features are often created during feature engineering. insert() can add a calculated feature at a chosen position in the dataset. The new feature can then be inspected, transformed, or used during model training.

# Explanation
df.insert() modifies the DataFrame by adding a new column at the specified integer position. The first argument is the position, the second is the new column name, and the third contains the values. The number of supplied values must match the number of rows.

# Python Implementation

```python
insert_df = df.copy()

# Insert Grade column at position 3
insert_df.insert(
    3,
    "Grade",
    ["A", "B", "A", "A", "C", "A", "B", "A"]
)

print(insert_df)
```

# 5. assign()
Definition
The assign() method creates new columns and returns a new DataFrame. It is useful when we want to add calculated or transformed columns without directly changing the original DataFrame. Multiple new columns can also be created in one operation.

# Real-Life Example
A company may calculate a Bonus column from employee salary data. assign() can create the new calculated column while preserving the original dataset. This makes it convenient to create different versions of business data.

# AI/ML Example
In AI/ML, assign() can create new features from existing features during feature engineering. For example, a percentage, ratio, or derived measurement can be added as a new column. These derived features may provide useful information to a machine learning model.

# Explanation
df.assign() returns a new DataFrame containing the additional column. Existing columns can be used inside the calculation for the new feature. Because it returns a new DataFrame, the result should normally be stored in a variable if we want to use it later.

# Python Implementation

```python
# Create a new Percentage column
assigned_df = df.assign(
    Percentage=df["Marks"]
)

print(assigned_df)
```

# 6. apply()
Definition
The apply() method applies a function to values along a DataFrame column or across rows and columns. It is useful when a custom transformation cannot be expressed as a simple built-in operation. Functions or lambda expressions can be used with apply().

# Real-Life Example
A college may convert numerical marks into result categories such as Pass or Fail. apply() can run the same decision function for every student's mark. This avoids manually checking each record.

# AI/ML Example
In AI/ML, apply() is often used during feature engineering and data preprocessing. It can transform raw feature values into categories or calculate custom features. However, vectorized Pandas operations are generally preferred when they can perform the same task efficiently.

# Explanation
When apply() is used on a Series, the supplied function is applied to each value. In this example, a lambda function checks each Marks value and returns Pass or Fail. The returned values are stored in a new column.

# Python Implementation

```python
apply_df = df.copy()

apply_df["Result"] = apply_df["Marks"].apply(
    lambda mark: "Pass" if mark >= 75 else "Fail"
)

print(apply_df)
```

# 7. map()
Definition
The map() method transforms values in a Pandas Series using a dictionary, function, or another Series. It is commonly used to convert categorical values into new labels or numerical representations. map() is mainly used with a single Series or column.

# Real-Life Example
A company may store employee status as Active and Inactive but wants to represent them as 1 and 0 for analysis. map() can convert each category according to a predefined mapping. This makes categorical information easier to process.

# AI/ML Example
In AI/ML, map() can perform simple categorical encoding before model training. For example, Yes and No values can be converted into 1 and 0. It is useful when the categories have a clear and known mapping.

# Explanation
Series.map() checks each value and replaces it according to the supplied dictionary or function. Values that are not found in a mapping dictionary may become NaN. Therefore, the mapping should include all expected categories when appropriate.

# Python Implementation

```python
map_df = df.copy()

map_df["City_Code"] = map_df["City"].map({
    "Bengaluru": 1,
    "Mysuru": 2,
    "Hassan": 3
})

print(map_df)
```

# 8. replace()
Definition
The replace() method substitutes specified values with new values in a Series or DataFrame. It can replace one value, multiple values, or values based on a mapping. It is useful for correcting or standardizing data.

# Real-Life Example
A company dataset may contain the abbreviation BLR for Bengaluru. An analyst can use replace() to change BLR into Bengaluru so location names follow a consistent format. Standardized values make reports more reliable.

# AI/ML Example
In AI/ML, inconsistent categories can create problems during preprocessing and encoding. replace() can standardize different representations of the same category before model training. It can also replace specific numeric or categorical values when cleaning data.

# Explanation
df.replace() searches for specified values and substitutes them with the new values. A dictionary can be used when several replacements are needed. The operation can be applied to the complete DataFrame or to a specific column.

# Python Implementation

```python
replace_df = df.copy()

replace_df["City"] = replace_df["City"].replace({
    "Bengaluru": "BLR",
    "Karur": "KRR"
})

print(replace_df)
```

# 9. Duplicate Handling
Definition
Duplicate handling means identifying and removing repeated records from a dataset. Pandas provides duplicated() to detect duplicate rows and drop_duplicates() to remove them. Managing duplicates helps maintain accurate and consistent data.

# Real-Life Example
A company may accidentally import the same customer transaction more than once. duplicated() can identify repeated records, and drop_duplicates() can remove them. This prevents duplicate transactions from affecting business reports.

# AI/ML Example
In AI/ML, duplicate samples can bias analysis and model training because repeated observations may receive unnecessary importance. Detecting and removing unintended duplicates improves dataset quality. However, genuine repeated records should not be removed without checking their meaning.

# Explanation
df.duplicated() returns True for rows considered duplicates. df.drop_duplicates() removes repeated rows and keeps the first occurrence by default. We can also use the subset parameter when duplicates should be checked using only selected columns.

# Python Implementation

```python
duplicate_df = pd.concat([df, df.iloc[[0]]], ignore_index=True)

# Detect duplicate rows
print(duplicate_df.duplicated())

# Remove duplicate rows
clean_df = duplicate_df.drop_duplicates()

print(clean_df)
```
